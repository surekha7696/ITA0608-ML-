```python

import csv

# Create CSV file
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

with open('data.csv', 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['Sky', 'Temp', 'Humidity', 'Wind', 'Water', 'Forecast', 'Target'])
    w.writerows(data)

# Read CSV file
with open('data.csv') as f:
    rows = list(csv.reader(f))[1:]

n = len(rows[0]) - 1
S = ['0'] * n
G = [['?'] * n]

for row in rows:
    x, y = row[:-1], row[-1]

    if y == 'Yes':
        G = [g for g in G if all(g[i] == '?' or g[i] == x[i] for i in range(n))]
        for i in range(n):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []
        for g in G:
            if all(g[i] == '?' or g[i] == x[i] for i in range(n)):
                for i in range(n):
                    if g[i] == '?' and S[i] not in ('0', '?') and S[i] != x[i]:
                        h = g.copy()
                        h[i] = S[i]
                        new_G.append(h)
            else:
                new_G.append(g)
        G = new_G

print("Specific Boundary:", S)
print("General Boundary:")
for g in G:
    print(g)
```
